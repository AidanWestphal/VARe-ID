```python
import json
from collections import defaultdict
import os
from PIL import Image
import matplotlib.pyplot as plt
import math
import numpy as np

# --- Step 1: Specify the path to your JSON file ---
json_file_path = '/fs/ess/PAS2136/ggr_data/results/Grevys_Experiment_10_IA95_vp_limited/Grevys_Exp_10_IA80_vp_limited/lca/lca_right_annots.json'
json_out_dir = '/fs/ess/PAS2136/ggr_data/results/Grevys_Experiment_10_IA95_vp_limited/Grevys_Exp_10_IA80_vp_limited/lca/cluster_images_right'

os.makedirs(json_out_dir, exist_ok=True)

# --- Configuration ---
IMAGES_PER_LINE = 5
FIELDS_TO_PRINT = ['uuid', 'image_uuid', 'tracking_id', 'CA_score']

# --- Step 2: Load and process the data ---
try:
    with open(json_file_path, 'r') as f:
        data = json.load(f)

    print(f"Successfully loaded and processing: '{json_file_path}'\n")

    annotations = data.get('annotations', [])
    image_info = {img['uuid']: img for img in data.get('images', [])}

    clusters = defaultdict(list)
    for annotation in annotations:
        cluster_id = annotation.get('LCA_clustering_id')
        if cluster_id is not None:
            clusters[cluster_id].append(annotation)

    # --- Step 3: Iterate through each cluster and create a grid display ---
    # --- THE FIX ---
    # This sorts by the number before the underscore, ignoring the rest.
    for cluster_id in sorted(clusters.keys(), key=lambda x: int(x.split('_')[0])):
        annotations_in_cluster = clusters[cluster_id]
        num_annotations = len(annotations_in_cluster)
        
        # Calculate grid size
        num_rows = math.ceil(num_annotations / IMAGES_PER_LINE)
        
        fig, axes = plt.subplots(num_rows, IMAGES_PER_LINE, figsize=(4 * IMAGES_PER_LINE, 5 * num_rows))
        
        # --- THE FIX: This robustly handles the 'axes' object ---
        # It ensures that no matter what plt.subplots returns (a single object, a 1D array, or a 2D array),
        # we always end up with a simple, flat list of plot axes to work with.
        if isinstance(axes, np.ndarray):
            axes_flat = axes.flatten()
        else:
            axes_flat = [axes] # Make a single Axes object into a list

        print(f"--- Cluster ID: {cluster_id} ({num_annotations} annotations) ---")
        
        for i, (ax, ann) in enumerate(zip(axes_flat, annotations_in_cluster)):
            image_uuid = ann.get('image_uuid')
            if image_uuid in image_info:
                img_path = image_info[image_uuid].get('image_path')
                try:
                    with Image.open(img_path) as full_image:
                        bbox = ann['bbox']
                        x, y, w, h = bbox
                        cropped_image = full_image.crop((x, y, x + w, y + h))
                        ax.imshow(cropped_image)
                except (FileNotFoundError, TypeError):
                    ax.text(0.5, 0.5, 'Image not found', ha='center', va='center')

            ax.set_title(f"Annotation {i+1}")
            ax.set_xticks([])
            ax.set_yticks([])

            # Create the text block for display
            info_lines = []
            for key in FIELDS_TO_PRINT:
                if key == 'CA_score':
                    value = f"{ann.get(key, 0.0):.4f}"
                else:
                    value = ann.get(key, 'N/A')
                info_lines.append(f"- {key}: {value}")
            
            img_path_short = os.path.basename(image_info.get(image_uuid, {}).get('image_path', 'N/A'))
            info_lines.append(f"- image_file: {img_path_short}")

            ax.set_xlabel('\n'.join(info_lines), fontsize=9, ha='center', va='top')

        # Hide any unused subplots
        for i in range(num_annotations, len(axes_flat)):
            axes_flat[i].axis('off')

        plt.tight_layout(pad=3.0)

        out_fpath = os.path.join(json_out_dir, f"{cluster_id}.jpg")
        plt.savefig(out_fpath)
        
        plt.clf()

except FileNotFoundError:
    print(f"Error: The file was not found at '{json_file_path}'")
except Exception as e:
    print(f"An error occurred: {e}")
```
